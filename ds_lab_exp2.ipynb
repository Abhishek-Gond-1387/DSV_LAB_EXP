```python
#Q 8.2
counter = 0
def show_counter():
    print("Counter =", counter)
show_counter()
```

```python
#Q 8.3
counter = 0

def increment_counter():
    counter = counter + 1
    print(counter)

increment_counter()
```

UnboundLocalError: cannot access local variable 'counter' where it is not associated with a value

When python sees: counter = counter + 1 it assumes counter is a local variable because you are assigning a value to it.

However, before assigning, it tries to evaluate the right-hand side: counter + 1
Since the local counter has not been given any value yet, Python raises an UnboundLocalError.
