```python
#Q35 Examination performance analysis: Generate a reproducible 12 x 4 NumPy array of integer test marks from 20 to 100, inclusive, using a fixed seed. Calculate each student's average and each test's mean and standard deviation. Count students passing all four tests (at least 40 in each). Identify students with averages of at least 75 and the test with greatest variation. State your student numbering convention.

import numpy as np

np.random.seed(42)

marks = np.random.randint(20, 101, (12, 4))

student_avg = marks.mean(axis=1)
test_mean = marks.mean(axis=0)
test_std = marks.std(axis=0)

passing = np.all(marks >= 40, axis=1)
high_average = np.where(student_avg >= 75)[0] + 1

print("Marks:\n", marks)
print("Student averages:", student_avg)
print("Test means:", test_mean)
print("Test standard deviations:", test_std)
print("Students passing all tests:", np.sum(passing))
print("Students with averages >= 75:", high_average)
print("Test with greatest variation:", np.argmax(test_std) + 1)

print("Students and tests are numbered starting from 1.")
```

```python
#Q36 Matrix transformation and analysis: Construct a 5 x 5 matrix containing integers 21 to 45. Extract the central 3 x 3 matrix, transpose it, and calculate the matrix product of the extracted matrix with its transpose. Calculate row-wise sums of the product and identify its maximum and all positions tied at that maximum. Explain dimensions at every stage.

import numpy as np

a = np.arange(21, 46).reshape(5, 5)
central = a[1:4, 1:4]
transposed = central.T
product = central @ transposed
row_sums = product.sum(axis=1)

maximum = row_sums.max()
positions = np.where(row_sums == maximum)[0]

print("Original matrix (5 x 5):\n", a)
print("Central matrix (3 x 3):\n", central)
print("Transpose (3 x 3):\n", transposed)
print("Product (3 x 3):\n", product)
print("Row-wise sums:", row_sums)
print("Maximum row sum:", maximum)
print("Positions tied at maximum (zero-based):", positions)
print("The 5 x 5 matrix becomes 3 x 3 after extraction.")
print("Multiplying a 3 x 3 matrix by its 3 x 3 transpose gives a 3 x 3 matrix.")
```

```python
#Q37 Sales and revenue analysis: Create a 6 x 4 matrix of quantities sold over six days for four products. Use unit prices [150, 240, 90, 320]. Calculate daily revenue, total revenue per product, the highest-revenue day, and the best-selling product by quantity. Reduce prices by 10% for products whose total quantity exceeds 100, then recalculate total revenue without modifying original prices.

import numpy as np

quantities = np.array([
    [10, 15, 20, 8],
    [12, 18, 25, 10],
    [15, 20, 22, 12],
    [18, 16, 30, 15],
    [20, 22, 28, 18],
    [25, 24, 35, 20]
])

prices = np.array([150, 240, 90, 320])

daily_revenue = quantities @ prices
product_revenue = quantities.sum(axis=0) * prices
total_quantity = quantities.sum(axis=0)

new_prices = np.where(total_quantity > 100, prices * 0.9, prices)
new_daily_revenue = quantities @ new_prices

print("Daily revenue:", daily_revenue)
print("Total revenue per product:", product_revenue)
print("Highest-revenue day:", np.argmax(daily_revenue) + 1)
print("Best-selling product:", np.argmax(total_quantity) + 1)
print("Total quantity per product:", total_quantity)
print("Original prices:", prices)
print("Reduced prices:", new_prices)
print("Revenue after price reduction:", new_daily_revenue.sum())
```

```python
#Q38 Random data cleaning and statistics: Generate 100 normal observations with mean 50 and standard deviation 12 using a fixed seed. Count values outside 30-70 and replace them with the nearest boundary using np.where(). Compare original and cleaned mean, variance, standard deviation, minimum, and maximum. Explain the statistical effects of boundary replacement.

import numpy as np

np.random.seed(42)

data = np.random.normal(50, 12, 100)
cleaned = np.where(data < 30, 30, np.where(data > 70, 70, data))

print("Values outside 30-70:", np.sum((data < 30) | (data > 70)))

print("Original mean:", np.mean(data))
print("Cleaned mean:", np.mean(cleaned))

print("Original variance:", np.var(data))
print("Cleaned variance:", np.var(cleaned))

print("Original standard deviation:", np.std(data))
print("Cleaned standard deviation:", np.std(cleaned))

print("Original minimum:", np.min(data))
print("Cleaned minimum:", np.min(cleaned))

print("Original maximum:", np.max(data))
print("Cleaned maximum:", np.max(cleaned))

print("Boundary replacement limits extreme values and generally reduces variance and standard deviation.")
print("It may also change the mean and restricts the minimum and maximum to the range 30-70.")
```

```python
#Q39 Multi-branch financial reconciliation: Combine the three earnings Series by ordinary addition. Explain missing values, identify quarters common to all branches, and calculate their combined earnings. Construct a separate result treating absent branch records as zero and explain how its interpretation differs.

import pandas as pd

branch_a = pd.Series({"Q1": 120, "Q2": 145, "Q4": 180})
branch_b = pd.Series({"Q2": 110, "Q3": 135, "Q4": 155})
branch_c = pd.Series({"Q1": 95, "Q3": 125, "Q4": 140})

result = branch_a + branch_b + branch_c
print("Ordinary addition:\n", result)

common = branch_a.index.intersection(branch_b.index).intersection(branch_c.index)
combined = branch_a[common] + branch_b[common] + branch_c[common]

print("Quarters common to all branches:", common.tolist())
print("Combined earnings for common quarters:\n", combined)

zero_result = branch_a.add(branch_b, fill_value=0).add(branch_c, fill_value=0)
print("Addition treating absent records as zero:\n", zero_result)

print("Ordinary addition produces NaN when a branch lacks a quarter.")
print("Treating absent records as zero assumes that missing records represent zero earnings.")
```

```python
#Q40 Series data integrity: Preserve the required index order, identify missing student records, and filter valid scores of at least 70. Convert the filtered Series to a dictionary, reconstruct it as a Series, and verify that labelled values match. Explain why missing records should not automatically be zero marks.

import pandas as pd

scores = {"R101": 78, "R102": 65, "R104": 92, "R105": 48}
required = ["R105", "R103", "R101", "R106", "R104", "R102"]

s = pd.Series(scores, index=required)

print("Original Series:\n", s)
print("Missing student records:", s[s.isnull()].index.tolist())

valid = s[s.notnull() & (s >= 70)]
print("Valid scores >= 70:\n", valid)

data = valid.to_dict()
reconstructed = pd.Series(data)

print("Dictionary:", data)
print("Reconstructed Series:\n", reconstructed)
print("Labelled values match:", valid.equals(reconstructed))

print("Missing records should not automatically be assigned zero because no score may have been recorded.")
```
